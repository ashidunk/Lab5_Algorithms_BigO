```python
# APPENDIX 1: EMERGENCY DISPATCH RECORD SEARCH

import time

# Student-specific inputs
LAST_NAME = "Sabariaga"
STUDENT_ID = "TUPM-26-1862"
FAVORITE_ARTIST = "MICO"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(
    ord(c.upper()) - ord("A") + 1
    for c in FAVORITE_ARTIST
    if c.isalpha()
)

# Generate unique incident records
incident_records = sorted({
    SEED_DIGIT * 10 + ID_SUM + i * 3
    for i in range(15)
})

# Rotate the records using SEED_DIGIT
rotation = SEED_DIGIT % len(incident_records)
rotated_records = (
    incident_records[rotation:]
    + incident_records[:rotation]
)

# Generate the target incident
target_incident = ID_SUM + ARTIST_SUM

print("=== EMERGENCY DISPATCH RECORD SEARCH ===")
print("Original Dataset:", incident_records)
print("Rotated Dataset:", rotated_records)
print("Rotation Point:", rotation)
print("Target Incident:", target_incident)


# Search a rotated sorted list using modified Binary Search
def rotated_binary_search(data, target):
    low, high = 0, len(data) - 1
    comparisons = 0

    while low <= high:
        mid = (low + high) // 2
        comparisons += 1

        if data[mid] == target:
            return mid, comparisons

        # Determine which half is sorted
        if data[low] <= data[mid]:
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1

    return -1, comparisons


# Linear Search for comparison
def linear_search(data, target):
    comparisons = 0

    for index, value in enumerate(data):
        comparisons += 1

        if value == target:
            return index, comparisons

    return -1, comparisons


binary_index, binary_comparisons = rotated_binary_search(
    rotated_records, target_incident
)

linear_index, linear_comparisons = linear_search(
    rotated_records, target_incident
)

print("\n=== BINARY SEARCH RESULT ===")
if binary_index != -1:
    print("Incident Found at Index:", binary_index)
    print("Incident Position:", binary_index + 1)
else:
    print("Incident Not Found")

print("Binary Search Comparisons:", binary_comparisons)

print("\n=== LINEAR SEARCH RESULT ===")
if linear_index != -1:
    print("Incident Found at Index:", linear_index)
    print("Incident Position:", linear_index + 1)
else:
    print("Incident Not Found")

print("Linear Search Comparisons:", linear_comparisons)
print(
    "Comparison Difference:",
    linear_comparisons - binary_comparisons
)

print("\n=== COMPLEXITY ANALYSIS ===")
print("Linear Search Time: O(n)")
print("Binary Search Time: O(log n) average for this rotated-array method")
print("Auxiliary Space: O(1) for both search functions")
```

```python
# APPENDIX 3: E-COMMERCE ORDER PROCESSING

import random
import time

# Student-specific inputs
LAST_NAME = "Sabariaga"
STUDENT_ID = "TUPM-26-1862"
FAVORITE_ARTIST = "MICO"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(
    ord(c.upper()) - ord("A") + 1
    for c in FAVORITE_ARTIST
    if c.isalpha()
)

# Generate student-specific order numbers
order_count = 10

original_orders = [
    (ID_SUM + ARTIST_SUM + i * SEED_DIGIT + i * 7)
    for i in range(order_count)
]

# Create three input conditions
sorted_orders = sorted(original_orders)
reverse_orders = sorted(original_orders, reverse=True)

random_orders = original_orders.copy()
random.Random(SEED_DIGIT + ID_SUM).shuffle(random_orders)

datasets = {
    "Sorted": sorted_orders,
    "Reverse": reverse_orders,
    "Random": random_orders
}

print("=== GENERATED ORDER DATASETS ===")
print("Dataset Size:", order_count)
print("Sorted Orders:", sorted_orders)
print("Reverse-Ordered Orders:", reverse_orders)
print("Random Orders:", random_orders)


# Bubble Sort
def bubble_sort(data):
    a = data.copy()
    comparisons = 0
    movements = 0

    for i in range(len(a)):
        swapped = False

        for j in range(len(a) - i - 1):
            comparisons += 1

            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                movements += 1
                swapped = True

        if not swapped:
            break

    return a, comparisons, movements


# Selection Sort
def selection_sort(data):
    a = data.copy()
    comparisons = 0
    movements = 0

    for i in range(len(a) - 1):
        minimum = i

        for j in range(i + 1, len(a)):
            comparisons += 1

            if a[j] < a[minimum]:
                minimum = j

        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            movements += 1

    return a, comparisons, movements


# Insertion Sort
def insertion_sort(data):
    a = data.copy()
    comparisons = 0
    movements = 0

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if a[j] <= key:
                break

            a[j + 1] = a[j]
            movements += 1
            j -= 1

        a[j + 1] = key

    return a, comparisons, movements


algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

print("\n=== SORTING PERFORMANCE COMPARISON ===")

for condition, orders in datasets.items():
    print(f"\n--- {condition} Input ---")

    for name, algorithm in algorithms.items():
        start = time.perf_counter()

        sorted_result, comparisons, movements = algorithm(orders)

        elapsed = time.perf_counter() - start

        print(f"\nAlgorithm: {name}")
        print("Comparisons:", comparisons)
        print("Movements:", movements)
        print(f"Execution Time: {elapsed:.8f} seconds")
        print("Sorting Correct:", sorted_result == sorted(orders))


# Empirical performance: repeat each test five times
print("\n=== EMPIRICAL PERFORMANCE ===")
print("Dataset Size:", order_count)
print("Number of Trials: 5")

for condition, orders in datasets.items():
    print(f"\n{condition} Input:")

    for name, algorithm in algorithms.items():
        times = []

        for _ in range(5):
            start = time.perf_counter()
            algorithm(orders)
            elapsed = time.perf_counter() - start
            times.append(elapsed)

        average_time = sum(times) / len(times)

        print(
            f"{name}: Average Execution Time = "
            f"{average_time:.8f} seconds"
        )


print("\n=== BIG O ANALYSIS ===")
print("Bubble Sort: Best O(n) with early stopping; average/worst O(n^2)")
print("Selection Sort: Best, average, and worst O(n^2)")
print("Insertion Sort: Best O(n); average/worst O(n^2)")
print("Auxiliary Space: O(n) for each implementation because it copies the input")

print("\n=== FINAL VERIFICATION ===")
for condition, orders in datasets.items():
    print(f"{condition} Input:")

    for name, algorithm in algorithms.items():
        result, _, _ = algorithm(orders)
        print(
            f"{name}:",
            "PASS" if result == sorted(orders) else "FAIL"
        )
```

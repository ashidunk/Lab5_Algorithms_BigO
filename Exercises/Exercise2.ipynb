```python
# APPENDIX 2: UNIVERSITY ENROLLMENT RECORDS

# Student-specific inputs
LAST_NAME = "Sabariaga"
STUDENT_ID = "TUPM-26-1862"
FAVORITE_ARTIST = "MICO"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)

# Generate two ascending enrollment datasets
dataset_a = sorted({
    SEED_DIGIT * 10 + ID_SUM + i * 4
    for i in range(10)
})

dataset_b = sorted({
    ARTIST_LENGTH * 10 + NAME_LENGTH + i * 5
    for i in range(10)
})

print("=== UNIVERSITY ENROLLMENT RECORDS ===")
print("Department A Dataset:", dataset_a)
print("Department B Dataset:", dataset_b)


# Merge two sorted datasets and count comparisons
def merge_sorted(a, b):
    i = 0
    j = 0
    comparisons = 0
    merged = []

    while i < len(a) and j < len(b):
        comparisons += 1

        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1

    # Add remaining records
    merged.extend(a[i:])
    merged.extend(b[j:])

    return merged, comparisons


consolidated, comparison_count = merge_sorted(
    dataset_a, dataset_b
)

print("\n=== CONSOLIDATED ENROLLMENT DATA ===")
print("Consolidated Dataset:", consolidated)

print("\n=== RECORD COUNT VERIFICATION ===")
print("Department A Records:", len(dataset_a))
print("Department B Records:", len(dataset_b))
print("Records Before Consolidation:", len(dataset_a) + len(dataset_b))
print("Records After Consolidation:", len(consolidated))

print("\n=== ORDERING VERIFICATION ===")
print(
    "Ascending Order:",
    consolidated == sorted(consolidated)
)
print(
    "All Records Preserved:",
    len(consolidated) == len(dataset_a) + len(dataset_b)
)

print("\n=== PERFORMANCE ANALYSIS ===")
print("Merge Comparisons:", comparison_count)
print("Time Complexity: O(n + m)")
print("Auxiliary Space Complexity: O(n + m)")
```
